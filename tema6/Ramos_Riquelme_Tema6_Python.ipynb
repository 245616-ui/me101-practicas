```python
import matplotlib.pyplot as plt
import numpy as np

fila = np.linspace(0, 255, 5)
gris = np.tile(fila, (5, 1)).astype(np.uint8)

print("Shape:", gris.shape)
print("Dtype:", gris.dtype)
print("Mean:", gris.mean())
```

```python
plt.imshow(gris, cmap="gray", vmin=0, vmax=250)
plt.colorbar()
plt.title("Degradado en Escala de Grises")
plt.savefig("gris_tema6.png")
plt.show()
```

```python
recorte = gris[:, 3:]
print("Recorte mitad derecha:\n", recorte)
```

```python
rgb = np.zeros((4, 4, 3), dtype=np.uint8)
rgb[:, 0:2] = [0, 255, 0]
rgb[:, 2:4] = [0, 0, 255]

print("Shape RGB:", rgb.shape)
print("Pixel [0,0] (Verde):", rgb[0, 0])
print("Pixel [0,3] (Azul):", rgb[0, 3])
```

```python
canal_R = rgb[:, :, 0]
canal_G = rgb[:, :, 1]
canal_B = rgb[:, :, 2]
```

```python
gris_promedio = np.mean(rgb, axis=2)
gris_luminosidad = (
    0.299 * rgb[:, :, 0] + 0.585 * rgb[:, :, 1] + 0.114 * rgb[:, :, 2]
)

print("\nGris Promedio:\n", gris_promedio)
print("\nGris Luminosidad:\n", gris_luminosidad)

import matplotlib.pyplot as plt


fig, axes = plt.subplots(1, 2, figsize=(10, 5))

axes[0].imshow(gris_promedio, cmap="gray", vmin=0, vmax=255)
axes[0].set_title("Gris Promedio")
axes[0].axis("off")


axes[1].imshow(gris_luminosidad, cmap="gray", vmin=0, vmax=255)
axes[1].set_title("Gris Luminosidad")
axes[1].axis("off")

plt.tight_layout()
plt.show()
```

```python
umbral = 100
binaria = gris_luminosidad > umbral
print("\nMatriz Binarizada (Umbral > 100):\n", binaria)
import matplotlib.pyplot as plt


plt.imshow(binaria, cmap="gray", vmin=0, vmax=1)
plt.title(f"Matriz Binarizada (Umbral > {umbral})")
plt.axis("off")
plt.show()
```
